```python
import faiss

import numpy as np

import os
```

```python
os.makedirs(
    "../faiss_index",
    exist_ok=True
)
```

```python
embeddings = np.load(
    "../embeddings/wiki_embeddings.npy"
)
```

```python
print(embeddings.shape)
```

```python
embeddings = embeddings.astype("float32")
```

```python
dimension = embeddings.shape[1]

print("Embedding dimension:", dimension)
```

```python
index = faiss.IndexFlatL2(dimension)
```

```python
index.add(embeddings)
```

```python
print("Total vectors in index:")

print(index.ntotal)
```

```python
faiss.write_index(
    index,
    "../faiss_index/wiki.index"
)
```

```python
print("FAISS index created successfully!")
```
