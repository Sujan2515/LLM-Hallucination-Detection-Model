```python
from sentence_transformers import SentenceTransformer

from datasets import load_from_disk

import numpy as np

import os
```

## Create Embeddings Folder

```python
os.makedirs(
    "../embeddings",
    exist_ok=True
)
```

```python
wiki = load_from_disk(
    "../datasets/Wikipedia"
)
```

```python
print(wiki)
```

```python
print(wiki[0])
```

```python
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
wiki_texts = []

CHUNK_SIZE = 300

for sample in wiki:

    text = sample["text"]

    if not text:
        continue

    # Split article into chunks
    for i in range(0, len(text), CHUNK_SIZE):

        chunk = text[i:i + CHUNK_SIZE]

        # skip tiny chunks
        if len(chunk.strip()) < 100:
            continue

        wiki_texts.append(chunk)

print("Total chunks:", len(wiki_texts))
```

```python
print("Total Wikipedia articles:")

print(len(wiki_texts))
```

## Generate embeddings

```python
all_embeddings = []

batch_size = 512

for i in range(0, len(wiki_texts), batch_size):

    batch = wiki_texts[i:i + batch_size]

    batch_embeddings = model.encode(
        batch,
        show_progress_bar=True
    )
    
    all_embeddings.append(batch_embeddings)

    print(f"Processed {i + len(batch)} / {len(wiki_texts)}")
```

```python
import pickle

with open(
    "../embeddings/wiki_chunks.pkl",
    "wb"
) as f:

    pickle.dump(
        wiki_texts,
        f
    )

print("Chunks saved successfully!")
```

```python
embeddings = np.vstack(all_embeddings)
```

```python
print(embeddings.shape)
```

```python
embeddings = embeddings.astype("float32")
```

```python
np.save(
    "../embeddings/wiki_embeddings.npy",
    embeddings
)
```
