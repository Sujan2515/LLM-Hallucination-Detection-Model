```python
from transformers import pipeline
```

```python
nli_pipeline = pipeline(
    "text-classification",
    model="MoritzLaurer/deberta-v3-base-mnli-fever-anli",
    device=0
)
```

```python
def verify_claim(claim, evidence):

    result = nli_pipeline(
        {
            "text": claim,
            "text_pair": evidence
        }
    )

    return result
```

```python
claim = "Paris is the capital of Germany"
```

```python
evidence = "Paris is the capital city of France."
```

```python
result = verify_claim(
    claim,
    evidence
)
```

```python
print(result)
```
